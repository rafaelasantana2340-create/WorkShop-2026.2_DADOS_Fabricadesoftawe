## 1.2 - Limpeza de dados

---



Agora que temos uma visão geral dos dados, é hora de limpá-los! Remova inconsistências, como dados ausentes e tipos incorretos, para garantir a qualidade da análise exploratória. Novamente, consulte a documentação do Pandas para ajudar nesse processo.

### Lembre-se que existem muitas funções ao seu favor! ✨

**dropna()**: Remove as linhas ou colunas com valores ausentes(nulos) ou irrelevantes para sua análise.

**fillna()**: Preenche os valores ausentes com algum valor específico, como a média, a mediana ou um valor constante, dependendo do contexto.

Essas funções são essenciais para garantir que seu dataset esteja limpo e pronto para análise! E o melhor é que existem muitas outras opções além dessas. Dêem uma olhada na documentação e descubram quais outras funções podem se adequar melhor à análise de vocês.

**Observação:**                                                     **df.isnull().sum():** ao verificar os valores nulos do dataset, foram encontrados dados faltantes nas colunas Age (177), Cabin (687) e Embarked (2). As demais colunas não apresentaram valores nulos.


```python
# Removendo as linhas que possuem valores ausentes na coluna Embarked
df = df.dropna(subset=['Embarked'])
```

**dropna():** utilizei essa função para remover os registros que não possuem informação na coluna **Embarked**. Foram identificados apenas 2 valores ausentes nessa coluna.

```python
# Preenchendo as idades ausentes com a mediana das idades
df['Age'] = df['Age'].fillna(df['Age'].median())
```

**fillna():** utilizei essa função para preencher os valores ausentes da coluna Age com a mediana das idades, evitando a remoção de vários registros do dataset.

```python
# Verificando os valores nulos após a limpeza dos dados
df.isnull().sum()
```

Após a limpeza dos dados, as colunas **Age e Embarked** não apresentam mais valores ausentes. A coluna Cabin ainda possui 687 valores nulos, sendo necessário avaliar seu tratamento devido à grande quantidade de informações faltantes.

```python
# Removendo a coluna Cabin, caso ela ainda exista no dataset
if 'Cabin' in df.columns:
    df = df.drop(columns=['Cabin'])
```

**drop():** utilizei essa função para remover a coluna Cabin, pois ela apresentou 687 valores ausentes, sendo a coluna com maior quantidade de dados não preenchidos.

```python
# Verificando os valores nulos após a limpeza dos dados
df.isnull().sum()
```

**Após realizar a limpeza dos dados, verifiquei novamente os valores nulos e todas as colunas apresentaram valor 0, indicando que não existem mais dados ausentes no dataset.**

## 1.3 - Análise Exploratória dos Dados (EDA)

---



🔍 Após concluir a etapa de limpeza dos dados, agora é o momento de analisar mais a fundo o que está acontecendo no dataset.

O próximo passo é identificar correlações e examinar a distribuição das variáveis presentes. Para isso, recomenda-se o uso da biblioteca [matplotlib](https://matplotlib.org/) para criar gráficos que representem essas relações.

Plote quantos gráficos forem necessários, não tenha pena, apenas bom senso!!

```python
# Analisando a quantidade de passageiros que sobreviveram ou não
df['Survived'].value_counts().plot(kind='bar')

# Adicionando título e identificação dos eixos
plt.title('Sobrevivência dos Passageiros')
plt.xlabel('Sobreviveu (0 = Não / 1 = Sim)')
plt.ylabel('Quantidade de Passageiros')

# Exibindo o gráfico
plt.show()
```

**Análise:** o gráfico compara a quantidade de passageiros que sobreviveram e que não sobreviveram ao acidente. O valor 0 representa os passageiros que não sobreviveram e o valor 1 representa os sobreviventes.

```python
# Comparando a sobrevivência dos passageiros por sexo
sobrevivencia_sexo = pd.crosstab(df['Sex'], df['Survived'])

# Criando o gráfico
sobrevivencia_sexo.plot(kind='bar')

# Adicionando título e identificação dos eixos
plt.title('Sobrevivência por Sexo')
plt.xlabel('Sexo')
plt.ylabel('Quantidade de Passageiros')
plt.legend(['Não sobreviveu', 'Sobreviveu'])

# Exibindo o gráfico
plt.show()
```

**Análise:** o gráfico mostra uma diferença na sobrevivência entre homens e mulheres. Entre as mulheres, a quantidade de sobreviventes foi maior que a de não sobreviventes. Já entre os homens, ocorreu o contrário, com uma quantidade maior de passageiros que não sobreviveram.

```python
# Comparando a sobrevivência dos passageiros por classe
sobrevivencia_classe = pd.crosstab(df['Pclass'], df['Survived'])

# Criando o gráfico
sobrevivencia_classe.plot(kind='bar')

# Adicionando título e identificação dos eixos
plt.title('Sobrevivência por Classe')
plt.xlabel('Classe do Passageiro')
plt.ylabel('Quantidade de Passageiros')
plt.legend(['Não sobreviveu', 'Sobreviveu'])

# Exibindo o gráfico
plt.show()
```

**Análise:** o gráfico mostra que na 1ª classe houve mais sobreviventes do que não sobreviventes. Na 2ª classe, os valores ficaram mais próximos. Já na 3ª classe, a quantidade de passageiros que não sobreviveram foi bem maior, indicando uma diferença na sobrevivência entre as classes.

```python
# Analisando a distribuição das idades dos passageiros
plt.hist(df['Age'], bins=20, edgecolor='black')

# Adicionando título e identificação dos eixos
plt.title('Distribuição das Idades dos Passageiros')
plt.xlabel('Idade')
plt.ylabel('Quantidade de Passageiros')

# Exibindo o gráfico
plt.show()
```

**Análise:** o histograma mostra uma maior concentração de passageiros adultos, principalmente entre aproximadamente 20 e 40 anos. Também existem passageiros mais jovens e mais idosos, porém em menor quantidade. O maior pico próximo dos 28 anos também está relacionado ao preenchimento dos valores ausentes da coluna Age com a mediana durante a limpeza dos dados.

```python
# Selecionando apenas as colunas numéricas do dataset
dados_numericos = df.select_dtypes(include=['number'])

# Calculando a correlação entre as variáveis numéricas
correlacao = dados_numericos.corr()

# Exibindo a matriz de correlação
correlacao
```

**Análise:** a tabela mostra os valores de correlação entre as variáveis numéricas do dataset, permitindo identificar quais possuem maior ou menor relação entre si.

## 1.4 - Feature engineering e dados categóricos

---



### 1.4.1 - Sempre que falamos de Data Science, lembramos dos diferentes tipos de dados. E, junto deles, surgem os temidos dados categóricos!

Esses dados são bem diferentes das variáveis numéricas, então a gente precisa dar uma atenção extra pra trabalhar com eles de forma eficaz. Bora transformar alguns deles usando os métodos que já vimos, como o **OneHotEncoder**! 🔥

Depois disso, vamos mergulhar no **Feature Engineering**!
Essa técnica é super poderosa e serve para extrair informações mais relevantes dos dados que já temos. Ela consiste em selecionar, manipular e transformar colunas e linhas do dataset, dando um novo significado ou enriquecendo as informações. Um exemplo clássico é a criação de uma coluna de IMC (Índice de Massa Corporal) a partir das colunas de altura e peso.

Agora é sua vez! Solte a criatividade e use sua intuição pra criar novos atributos a partir do que já está disponível.

**Dica**: Pegue as colunas que você já tem e use como base pra criar outras com características diferentes.

```python
# Importando o OneHotEncoder para transformar dados categóricos
from sklearn.preprocessing import OneHotEncoder
```

```python
# Criando o codificador
encoder = OneHotEncoder(sparse_output=False)

# Transformando a coluna Sex em valores numéricos
sexo_codificado = encoder.fit_transform(df[['Sex']])

# Visualizando os primeiros resultados
sexo_codificado[:5]
```

**OneHotEncoder:** utilizei esse método para transformar os dados categóricos da coluna Sex em valores numéricos, representados por 0 e 1.

```python
# Criando uma nova coluna com o tamanho da família de cada passageiro
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

# Visualizando os primeiros resultados
df[['SibSp', 'Parch', 'FamilySize']].head()
```

**Feature Engineering:** criei a coluna FamilySize a partir das colunas SibSp e Parch, para representar o tamanho da família de cada passageiro a bordo.

### 1.4.2 - Em poucos termos, computadores não conseguem entender texto de maneira alguma, mas números, ah, esses as máquinas entendem bem pra caramba! Então, que tal usar as técnicas que aprendemos pra deixar tudo necessário em números?

**Dica**: Sempre que temos uma feature categórica que possui apenas 2 elementos, podemos transformar ela em uma flag binária! Agora o que isso significa eu deixo pra você pesquisar 😉

```python
# Transformando a coluna Sex em uma flag binária
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# Visualizando os primeiros resultados
df[['Sex']].head()
```

**Flag binária:** transformei a coluna Sex em valores numéricos, utilizando 0 para male e 1 para female, facilitando o uso dessa informação na análise.
